```python
# =====================================================================
# TASK 2: DATA CLEANING
# =====================================================================
```

```python
print(df.isnull().sum())
```

```python
# Total count

print("\nTotal missing values in the dataset:", df.isnull().sum().sum())
```

```python
# Fix data types for next steps

num_cols = ['likes','comments','shares','watch_time_sec','impression_count','follower_count','engagement_rate','age']

for c in num_cols:
    df[c] = pd.to_numeric(df[c],errors = 'coerce')
```

```python
# Impute: median for numeric colums using 'fillna'

for c in num_cols:
    if df[c].isnull().sum()>0:
        df[c] = df[c].fillna(df[c].median())
```

```python
# Impute: mode for categorical colums using 'fillna'

for c in ['gender','sentiment','country','post_type','post_category','device_type']:
    if df[c].isnull().sum() > 0:
        df[c] = df[c].fillna(df[c].mode()[0])
```

```python
# Duplicate Handiling

#  Removing Duplicates

duplicate_count = df.duplicated().sum()
df = df.drop_duplicates()
print(f"Identified and removed {duplicate_count} absolute row duplicates.")
```

```python
print(df.isnull().sum())
```

```python
# Data formatting
```

```python
# Standardize categories using '.str,.strip(),tittle() and Lower()

df['gender'] = df['gender'].str.strip().str.title()
df['sentiment'] = df['sentiment'].str.strip().str.title()
df['country'] = df['country'].str.strip().str.title()
df['post_type'] = df['post_type'].str.strip().str.title()
df['device_type'] = df['device_type'].str.strip().str.title()

print(df[['gender', 'sentiment', 'country', 'post_type', 'device_type']].head(10))
```

```python
# Correct unrealistic values, here we remove the outlayers

for c in ['likes','comments','shares']:
    cap = df[c].quantile(0.99)
    df[c] = np.where(df[c] > cap, cap, df[c])
```

```python
#  Feature Extraction/Cleaning

# Extract hashtag count

df['hashtag_count'] = df['hashtags'].fillna('').str.findall(r'#\w+').str.len()

print(df[['hashtags', 'hashtag_count']].head(10))
print(df.shape)
```

```python
#cleane sentiment labels (their is only 3 types of unique values)

print(df['sentiment'].unique())
```

```python

# compute descriptive stats for [likes,comments,shares,watch_time_sec,engagement_rate,followers]
# mean,median,mode,standard deviation,variance and percentiles.
```

```python
print(df.columns)
```

```python
# Select the numeric columns
statistical_metrics = ['likes', 'comments', 'shares', 'engagement_rate','watch_time_sec','follower_count']

# creating a dataframe
statistical_summary = pd.DataFrame(index=['Mean', 'Median','mode' ,'Std Dev', 'Variance', '25th Percentile','50th Percentile', '75th Percentile','skewness','kurtosis'])

for col in statistical_metrics:
    statistical_summary[col] = [
        df[col].mean(),
        df[col].median(),
        df[col].mode()[0],
        df[col].std(),
        df[col].var(),
        df[col].quantile(0.25),
        df[col].quantile(0.50),
        df[col].quantile(0.75),
        df[col].skew(),
        df[col].kurt()
    ]
print(statistical_summary.round(2))
```

Task-5

Statistical analysis helps in understanding data by applying measures of central tendency and variability. Mean, median, and mode describe the typical values, while standard deviation and variance show how spread out the data is. Quartiles (Q1, Q2, and Q3) highlight the distribution range and help detect outliers. Skewness indicates whether the data is symmetrically distributed or leans to one side, while kurtosis shows how heavy or light the tails of the distribution are compared to a normal curve. Using these methods on metrics like likes, comments, shares, watch_time_sec, engagement_rate, and follower_count provides a clear picture of how people interact with social media posts.
